# Student Exam Score Data Processing

## Project objective
Load a raw CSV dataset, clean invalid data using reusable Python functions,
and analyze valid student exam scores using NumPy.

## Processing pipeline
1. Load the raw dataset.
2. Inspect and clean the records.
3. Convert valid scores into NumPy arrays.
4. Calculate descriptive statistics.
5. Review the results.


```python
import csv
import numpy as np
from pathlib import Path
```

```python
print(Path.cwd())
print(Path("data.csv").exists())
```

```python
def load_data(filename):
    # load csv rows
    with open(filename, "r", newline="") as file:
        return list(csv.DictReader(file))


data = load_data("data.csv")

print("Number of records:", len(data))
print("Column names:", list(data[0].keys()))
print("\nFirst three records:")

for row in data[:3]:
    print(row)
```

```python
def clean_data(data): 
    # remove duplicate and reject records with invalid scores. 
    cleaned = []
    rejected = []
    seen = set()

    for row in data: 
        student = row['student']

        # 1. skip exact duplicate records. 
        record_key = tuple(row.items())

        if record_key in seen: 
            continue
        seen.add(record_key)

        # 2. check scores are numeric
        try: 
            math_score = float(row["math_score"])
            python_score = float(row["python_score"])

        except ValueError:
            rejected.append((student, "missing or non-numeric score"))
            continue

        # 3. check scores are within the valid range
        if not(0 <= math_score <= 100 and 0 <= python_score <= 100):
            rejected.append((student, "score outside 0 - 100"))
            continue

        # 4. store valid scores as numbers. 
        cleaned.append({
            "student": student, 
            "math_score": math_score, 
            "python_score": python_score
        })

    return cleaned, rejected
```

```python
data = load_data("data.csv")
cleaned_data, rejected_data = clean_data(data) 

print("cleaned records:")
for row in cleaned_data:
    print(row)

print("\nRejected recors:")
for student, reason in rejected_data:
    print(f"{student}: {reason}")
```

```python
math_scores = np.array(
    [row["math_score"] for row in cleaned_data]
)

python_scores = np.array(
    [row["python_score"] for row in cleaned_data]
)

print("Math scores:", math_scores)
print("Python scores:", python_scores)
```

```python
math_scores = np.array(
    [row["math_score"] for row in cleaned_data]
)

python_scores = np.array(
    [row["python_score"] for row in cleaned_data]
)

print("Math scores:", math_scores)
print("Python scores:", python_scores)
```

```python
def analyze_data(score, subject): 
    # calculate statics using numpy
    print(f"\n{subject} statistics")
    print("-" * 25)
    print(f"Number of scores:", score.size)
    print(f"Mean: {np.mean(score):.2f}")
    print(f"Median: {np.median(score):.2f}")
    print(f"minimum: {np.min(score):.2f}")
    print(f"maximum: {np.max(score):.2f}")
    print(f"standard deviation: {np.std(score):.2f}")


analyze_data(math_scores, "Math")
analyze_data(python_scores, "Python")
```
